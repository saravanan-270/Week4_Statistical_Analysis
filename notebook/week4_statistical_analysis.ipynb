# Week 4 — Statistical Analysis and Hypothesis Testing
Gapminder country-level development dataset

```python
import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
```

```python
df=pd.read_csv('../data/gapminder_cleaned.csv')
d=df[df['year']==2007].copy()
print(d.shape)
```

## Hypothesis 1 — Africa vs Europe
**H0:** Mean life expectancy is equal.

**H1:** Mean life expectancy differs.

Test: Welch independent-samples t-test.

```python
af=d.loc[d.continent=='Africa','lifeexp']
eu=d.loc[d.continent=='Europe','lifeexp']
print('Africa mean:',af.mean())
print('Europe mean:',eu.mean())
print('Welch:',stats.ttest_ind(af,eu,equal_var=False))
print('Shapiro Africa:',stats.shapiro(af))
print('Shapiro Europe:',stats.shapiro(eu))
print('Levene:',stats.levene(af,eu,center='median'))
```

## Hypothesis 2 — GDP and life expectancy
**H0:** Pearson correlation is zero.

**H1:** Pearson correlation is not zero.

Test: Pearson correlation.

```python
print(stats.pearsonr(d['gdp_per_capita'],d['lifeexp']))
```

## Hypothesis 3 — Regression
**H0:** The log-GDP coefficient is zero.

**H1:** The log-GDP coefficient is not zero.

Test: OLS regression.

```python
X=sm.add_constant(d['log_gdp_per_capita'])
model=sm.OLS(d['lifeexp'],X).fit()
print(model.summary())
print('Shapiro residuals:',stats.shapiro(model.resid))
print('Breusch-Pagan:',het_breuschpagan(model.resid,model.model.exog))
```

## Hypothesis 4 — Continent and development group
**H0:** The variables are independent.

**H1:** The variables are associated.

Test: Chi-square test of independence.

```python
table=pd.crosstab(d['continent'],d['development_group'])
chi,p,dof,expected=stats.chi2_contingency(table)
print(table)
print('Chi-square:',chi,'p:',p,'df:',dof)
print('Minimum expected:',expected.min())
```

## Interpretation framework
Use alpha=0.05 as the primary threshold. Because four primary tests are conducted, also report a conservative Bonferroni threshold of 0.0125. Always report effect sizes and confidence intervals alongside p-values. Statistical association does not establish causation.
